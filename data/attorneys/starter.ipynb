# Attorneys by county in Ohio, California, Michigan, Idaho, Arizona, Washington and Pennsylvania, from each bar's public counts

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23777037](https://doi.org/10.5281/zenodo.23777037). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many lawyers each bar counts in each county, for seven states. Ohio: attorneys by county of address and status, from the Supreme Court of Ohio's attorney directory list (88 counties, with admission decade, private practice and Spanish speakers). California: active, inactive, judge and not-eligible licensees in each of the 58 counties, from the State Bar's county table. Michigan: active State Bar members living in each of the 83 counties, with private practice and the change from 2024. Idaho: members by status in each of the seven judicial districts. Arizona: active State Bar members in each of the 15 counties (the Bar's 2022 report). Washington: WSBA licensees of every type (lawyers plus limited licensees) in each of the 39 counties, with the attorney counts by member type. Pennsylvania: active attorneys in the seven counties the Disciplinary Board's annual report lists, and the Board's statewide status counts. Each record carries the source's own as-of date and a rate per 10,000 residents from the Census Bureau's county estimate. Only counts are published, and where a count is built from a directory file no name, birthdate, address or registration number is kept or written. In Ohio, counts of one to four in a disciplinary, deceased or other status, an admission decade, private practice or Spanish are withheld, and where a withheld count could be worked out from a county's total or the statewide figures, another count in the same group is withheld too. Every state record ties to its source's own total, and every published number is read a second time from the source file before release.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/attorneys/>
