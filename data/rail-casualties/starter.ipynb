# Railroad casualties and crossing incidents by county (FRA Forms 55a, 57 and 54)

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116057](https://doi.org/10.5281/zenodo.24116057). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Railroad casualties, highway-rail grade crossing incidents and train accidents that railroads reported to the Federal Railroad Administration (FRA), 2021 through July 2026, for 2,315 counties with at least one report, 50 states and the US: 41,307 casualties (5,136 deaths and 36,171 nonfatal injuries or illnesses) from Form 55a, split by who was hurt (railroad employee on duty or off duty, contractor or volunteer, passenger, trespasser, other non-trespasser), 12,345 crossing incidents with the people killed and injured (Form 57) and 14,018 rail equipment accidents (Form 54), with the casualty count of each incident year. FRA's county codes are Census FIPS codes; counties retired since keep their old code. Counts come from FRA's rolling six-year views on data.transportation.gov and are aggregated server-side; no narrative or person-level field is read.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/rail-casualties/>
