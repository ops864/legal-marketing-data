# Noncitizen population by state and county (Census ACS, table B05001)

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23850653](https://doi.org/10.5281/zenodo.23850653). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many residents of each state, DC and county are not U.S. citizens, with the margin of error and the share of the population, from the Census Bureau's American Community Survey 5-year estimates (2020-2024), table B05001 (nativity and citizenship status). Each record also splits the citizens into born in the US, born in a U.S. territory, born abroad of American parents and naturalized. These are survey estimates of residents, not of legal status: the table does not separate lawful permanent residents, visa holders or anyone else.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/noncitizen-population/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/noncitizen-population.csv"
URL = "https://probativegrowth.com/data/v1/noncitizen-population/noncitizen-population.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
