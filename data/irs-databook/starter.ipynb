Returns and other forms filed, gross tax collections and refunds issued, from the Internal Revenue Service Data Book state tables, for the nation, each state, DC and Puerto Rico and each fiscal year (October to September) from 2019 to 2025: returns by form type (individual, employment, C corporation, S corporation, partnership), collections by type of tax (individual income and employment, withheld, business income, estate, gift, excise) and the number and amount of refunds. Dollar amounts are in dollars; IRS prints them in thousands. The national record also holds what IRS counts in the nation but assigns to no state.

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/irs-databook.csv"
URL = "https://probativegrowth.com/data/v1/irs-databook/irs-databook.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
