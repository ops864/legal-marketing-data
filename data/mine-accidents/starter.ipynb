# Mine accidents, injuries and illnesses by state, county and degree of injury (MSHA), 2000 to 2026

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116045](https://doi.org/10.5281/zenodo.24116045). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Accident, injury and illness reports that U.S. mine operators and contractors file with the Mine Safety and Health Administration, counted for the nation and for each state and calendar year from 2000 to October 2026, and for each county over the last ten full years (2016 to 2025), by degree of injury: deaths, permanent disabilities, lost-time injuries, restricted-activity injuries, injuries with no lost time and all other reports. State and national records also split reports and deaths between coal and metal/nonmetal mines. Counts only: no narratives, names of operators, occupations or ages are kept.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/mine-accidents/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/mine-accidents.csv"
URL = "https://probativegrowth.com/data/v1/mine-accidents/mine-accidents.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
