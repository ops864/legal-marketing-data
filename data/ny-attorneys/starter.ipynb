# Attorneys registered with the New York courts, by county, status and year of admission (NYS Office of Court Administration)

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23753008](https://doi.org/10.5281/zenodo.23753008). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many lawyers are on New York's official attorney registration list, county by county. For each of the 62 counties the file counts the attorneys whose office address is there by registration status (currently registered, inside the 30-day reregistration window, delinquent, resigned, deceased, disbarred or suspended, incapacitated), by decade of admission to the New York bar, and per 10,000 residents using the Census Bureau's county population estimate. Further records count registrations with an office in each other state or territory, registrations with no office address in the public file, and the four judicial departments of admission, and the state record carries the admission-year history. Only counts are published: no name, address, phone number, firm or registration number is read, because the pull asks the court's open data portal for server-side totals and never downloads rows. Counts of one to four in a disciplinary status or an admission decade are withheld. Every record adds up to the state, the state ties to the source's own row count, and the American Bar Association's 2025 count of New York resident active lawyers sits between the two bounds the file allows.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/ny-attorneys/>
