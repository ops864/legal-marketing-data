New lawful permanent residents (green card holders) in the 200 U.S. counties with the most, for the two latest fiscal years: admissions by class (immediate relatives, family preference, employment preference, diversity, refugee and asylee adjustments, other) and the ten countries of birth with the most, with state sums set against the Yearbook of Immigration Statistics state totals. Read from the Department of Homeland Security's county workbooks, in which every cell is rounded to the nearest 10; sums are of rounded cells and understate the true figures, and the share of each state's total they cover is given.

## Load the data / Cargar los datos
