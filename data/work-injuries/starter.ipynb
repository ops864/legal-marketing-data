# Workplace injuries and fatal work injuries by state and industry (BLS SOII and CFOI)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23898154](https://doi.org/10.5281/zenodo.23898154). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How often workers are hurt on the job in each state and industry, and how many die: the nonfatal injury and illness rate per 100 full-time workers for private-industry sectors and detailed industries in 42 states and DC (Survey of Occupational Injuries and Illnesses, 2024), and the number and rate of fatal work injuries in every state and DC, with the major cause and the industry rate (Census of Fatal Occupational Injuries, 2024). Every record carries its BLS series or table. Survey estimates carry a relative standard error and a noise flag, and every figure is the one BLS publishes. It is statistics about industries and states, never about a named employer.

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/work-injuries.csv"
URL = "https://probativegrowth.com/data/v1/work-injuries/work-injuries.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
