# Crime and arrests by county and state: violent and property crime, DUI and drug arrests (FBI)

Starter notebook, version 2.1.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24011321](https://doi.org/10.5281/zenodo.24011321). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.1.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

What police agencies reported to the FBI over the last five complete calendar years, for every state and for 2,981 counties, home to 96.5% of U.S. residents: violent crimes, property crimes, arrests for driving under the influence, all arrests, and, for states and the United States, arrests for drug abuse violations, fraud, forgery and counterfeiting, and embezzlement. County rows add up the agencies the FBI places in each county (a city police department, a sheriff's office, a campus force; an agency that serves several counties counts in the county of its headquarters); a row says how many agencies there are and what share of the county's residents they cover, because each county counts only the months its agencies reported (never a zero in their place), and a county-year appears when at least one of its agencies reported. Rates are per 100,000 residents of the agencies that did report. State rows are the FBI's own state series; the county rows add up the agency-level answers, which can come to less than the state series where the FBI's state count includes reports its agency-level answers do not return. The United States row is the sum of the states. Only counts are published: no officer, victim, suspect or agency name.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/crime-data/>
