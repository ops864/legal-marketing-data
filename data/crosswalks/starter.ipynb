# Legal market crosswalks: practice areas, federal case codes, states, federal districts and counties

Starter notebook, version 2.3.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24181503](https://doi.org/10.5281/zenodo.24181503). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.3.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

The shared keys that join every Probative dataset. One record per practice area (our taxonomy of consumer and business practice areas, each linked to the FOLIO Area of Law class it matches, the federal nature-of-suit codes of its cases, the site's practice pages and the datasets that measure its demand, supply, rules and money), per federal nature-of-suit code, per state and territory (FIPS and postal codes, Census region and division, federal circuit and districts), per federal judicial district (our district key, the Federal Judicial Center code and the court's CM/ECF id), per county and county equivalent in the 50 states, DC and the territories (Puerto Rico's municipios and the county equivalents of Guam, the U.S. Virgin Islands, the Northern Mariana Islands, American Samoa and Midway: FIPS code, the federal district and statutory division it belongs to, its metro or micro area and its Census population), per former county code older sources still use (with the current counties that took its area), and per documented source code that names a place under another key (EEOC's WQ for Wake Island) or is not a place at all (military mail, the freely associated states, the VA's unplaced addresses, U.S. Courts' outside-the-U.S. code). Every county's district is read from the statute and checked against the counties U.S. Courts lists under each district in its bankruptcy Table F-5A.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/crosswalks/>
