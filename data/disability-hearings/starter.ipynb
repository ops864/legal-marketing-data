How long Social Security disability claimants wait for a hearing before an administrative law judge, office by office, from the Social Security Administration's own public data files: the average months from the hearing request to the hearing for cases closed in the month, and each hearing office's fiscal-year-to-date receipts, dispositions, cases pending and average days from request to decision, with SSA's rank among 159 offices.
