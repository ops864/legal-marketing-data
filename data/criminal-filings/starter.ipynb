# Federal criminal cases by offense and district (U.S. Courts Tables D-1, D-2 and D-3)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963315](https://doi.org/10.5281/zenodo.23963315). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Criminal defendants whose cases were opened in the 94 federal district courts, counted from the Statistical Tables for the Federal Judiciary that the Administrative Office of the U.S. Courts publishes for each 12-month period ending June 30 and December 31. Each record gives one district, one state with more than one district, or the whole country, for one period: the defendants commenced (transfers between districts left out), split by offense into nineteen columns and ten offense groups, with drug, immigration (illegal reentry shown on its own), firearms, fraud, sex offenses and traffic picked out, the felony, Class A misdemeanor and petty-offense split from Table D-1, and the same period a year earlier. National records add Table D-2's detail of about a hundred offenses, from bank robbery and health-care fraud to drunk driving on federal land. Every period's districts add up to the national total printed by the Courts, and every district ties to Table D-1.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/criminal-filings/>
