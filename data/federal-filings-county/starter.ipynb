# Federal civil cases filed, by county of the plaintiff's residence, case type and year (Federal Judicial Center)

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24024299](https://doi.org/10.5281/zenodo.24024299). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many federal civil lawsuits are filed in each U.S. county, and of what kind, over the last ten complete federal fiscal years, from the Federal Judicial Center's Integrated Database (IDB). A case is counted in the county where the first-listed plaintiff lives (the first-listed defendant's county when the United States sues; the tract's county in land condemnation cases), by 16 nature-of-suit groups plus multidistrict litigations (MDLs), and rolled up to Probative's practice areas. One record per county carries all ten years, the prior-year change and the year-by-year history. Cases that cannot be placed in a county are published as rows of their own, per state and nationally: plaintiffs who live in the U.S. outside the court's state (with the MDL share), plaintiffs who live abroad, cases under a retired county code, and cases with an unrecorded county; district records add the same parts for each of the 94 district courts. Every state equals the sum of its counties and these rows, and the national total ties to the U.S. Courts' Table C-2 within a fraction of a percent. Only counts are published: no party, docket number or judge.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/federal-filings-county/>
