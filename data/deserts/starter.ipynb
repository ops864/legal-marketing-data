# Legal deserts: law offices and lawyers per resident in every U.S. county

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963393](https://doi.org/10.5281/zenodo.23963393). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Which counties have few or no law offices for their residents, under two definitions shown side by side. Offices: private law-office establishments per 10,000 residents (BLS QCEW, latest quarter, 2025 population), flagged when under 1 or when there is none. Lawyers: where a state's own bar or court publishes a county count (California, New York, Mississippi, Illinois), lawyers per 1,000 residents, flagged when under 1, the threshold the American Bar Association used in 2020. Each county also shows the Census count of solo legal-services firms with no employees, and whether it is in a metropolitan area, a smaller city area or neither. State and national rows count the counties. Bar figures are county-by-status totals only; no lawyer is named.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/legal-deserts/>
