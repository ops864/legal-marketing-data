# Law firm size, solos and receipts, by state: Census SUSB 2022, Nonemployer Statistics 2024 and Economic Census 2022

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23898123](https://doi.org/10.5281/zenodo.23898123). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How big U.S. law firms are and what they take in, for the US, every state and DC, from three Census Bureau sources. Employer law firms (NAICS 541110, offices of lawyers) by number of employees (fewer than 5, 5 to 9, 10 to 19, 20 to 99, 100 to 499, 500 or more) with their employees, payroll and receipts, from the Statistics of U.S. Businesses (SUSB) 2022. Businesses with no employees (mostly sole practitioners) and their receipts, by legal form, from Nonemployer Statistics 2024, which publishes legal services (NAICS 5411) as one group. Firm counts, receipts, the share of receipts held by the largest firms and firms by receipts size from the Economic Census 2022. One record per jurisdiction and class. Every size-class row the Census Bureau publishes adds up to its total row, and every figure is the Bureau's own, never estimated.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/law-firm-size-and-revenue/>
