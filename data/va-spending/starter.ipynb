# VA disability compensation, pension and survivor dollars by county (VA GDX report)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23964263](https://doi.org/10.5281/zenodo.23964263). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Dollars the Department of Veterans Affairs paid in direct benefits in fiscal year 2025 in every U.S. county, from the Geographic Distribution of VA Expenditures (GDX) workbook of VA's National Center for Veterans Analysis and Statistics: disability compensation and veterans pension, dependency and indemnity compensation (DIC) and survivors pension, insurance, and specially adapted housing, with the total recomputed as the sum of the four (the workbook's Total column is a formula), and the compensation, pension and survivor dollars of the year before for the change. One record per county or county equivalent (3,223), per state, DC, Puerto Rico and Guam, for foreign countries, the Freely Associated States, other U.S. territories and unknown addresses, and for the US (187,660,338,257 dollars in all). Counties carry the Census FIPS code; Connecticut's counties are the planning regions.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/va-spending/>
