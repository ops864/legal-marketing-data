Workers' compensation claims assembled by the New York State Workers' Compensation Board in every New York county, for each accident year from 2000: the number of claims, how many have an attorney or representative (count and share), how many are controverted, how many have had a hearing and how many hearings were held, and how many are classified permanent partial or permanent total disability or occupational disease. Built only from SoQL aggregates of the Board's 5.7-million-claim file on NY Open Data (no claim-level row is read), with every count of 1 to 10 claims suppressed.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/ny-workers-comp/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/ny-workers-comp.csv"
URL = "https://probativegrowth.com/data/v1/ny-workers-comp/ny-workers-comp.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
