Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23898138](https://doi.org/10.5281/zenodo.23898138). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Veterans receiving VA disability compensation in every U.S. county on September 30, 2025 (fiscal year 2025), from the Department of Veterans Affairs county file on data.va.gov: the total and the count in each combined service-connected rating band (0 to 20%, 30 to 40%, 50 to 60%, 70 to 90% and 100%), by age group (17 to 44, 45 to 64, 65 and over) and by sex, with the same county's total and 100% count a year earlier. One record per county (3,144, in the 50 states and DC), per state and DC (the sum of its counties), for Puerto Rico, for the other territories, for veterans paid abroad, for unknown addresses and for the US. Every figure is one VA publishes; none is estimated.
