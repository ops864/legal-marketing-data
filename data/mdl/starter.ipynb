# Mass-tort docket tracker, federal MDL pending actions by month since 2019

Starter notebook, version 1.1.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23720737](https://doi.org/10.5281/zenodo.23720737). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.1.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Pending actions in every federal multidistrict litigation (MDL), from the Judicial Panel on Multidistrict Litigation's monthly 'Pending MDL Dockets by Actions Pending' reports. One record per docket: the court's docket title, the transferee district, the docket type, actions now pending, actions ever filed, the change over 12 months, the peak pending count, and the docket's monthly pending count since 2019. Version 1.1 adds the 14 MDLs the court closed between 2026-01-01 and 2026-10-01 (dateClosed set, the peak and last count from the monthly series) and the court's reports by district and by docket type, with monthly groupings of the series. Every report's rows add up exactly to the court's own totals row. The full monthly series, judge included, is at /data/v1/mdl/history/index.json (one CSV per year); the district, type and closed-MDL files are at /data/v1/mdl/reports/index.json. Counts and dates only.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/mass-tort-mdl-tracker/>
