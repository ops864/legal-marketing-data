# Recreational boating accidents by state and year: incidents, deaths, injuries and damage (U.S. Coast Guard), 2019 to 2024

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23962985](https://doi.org/10.5281/zenodo.23962985). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Recreational boating incidents reported to the U.S. Coast Guard, for the nation and for each state, DC and territory in each year from 2019 to 2024: incidents, fatal incidents, incidents with a non-fatal injury, property-damage-only incidents, deaths, people injured and property damage in dollars. The Coast Guard called these accidents until its 2024 report and now calls them incidents. Incidents offshore (Atlantic, Pacific, Gulf) are in the national record only. Each year comes from that year's Coast Guard report.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/boating-accidents/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/boating-accidents.csv"
URL = "https://probativegrowth.com/data/v1/boating-accidents/boating-accidents.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
