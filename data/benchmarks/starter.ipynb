Probative benchmarks, 2025–2026: our data on law-firm marketing. Figures are approximate and shown as ranges: bands, named tiers and rounded values, one table per file. The public tables cover how quickly inquiries become signed clients, phone calls and signed clients, AI answers and organic search clicks, search-ad click costs, what people type, calls from ads, Local Services Ads leads, inquiries by quarter and ad cost per signed client. Each row gives the measure, the band or tier, its unit and a rough size band. Read the band, not a point. The methods page defines every measure. Five injury-case tables were added in version 5.0.0 (each row states its version): Google Ads cost per click-traced lead and signed client, the non-brand click cost and the fourth-quarter dip, the Google Ads budget split, and Local Services Ads lead timing, charges and call lengths.
