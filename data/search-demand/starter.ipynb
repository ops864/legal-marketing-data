# Google searches for a lawyer by practice area and state, English and Spanish (Google Ads Keyword Planner, Trends and Autocomplete)

Starter notebook, version 1.3.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116059](https://doi.org/10.5281/zenodo.24116059). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.3.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many people search Google for a lawyer each month, practice by practice: 84 query families (personal injury and its niches such as truck accidents, mesothelioma and birth injury; immigration subtypes; bankruptcy chapters; estate planning; intellectual property; employment subtypes and more), each asked in the plain English and Spanish words people type. For the whole US, all 50 states and DC, the file gives the searches a month in English and in Spanish (12-month average), the head query of each language and the month-by-month series, from Google Ads Keyword Planner (October 2025 to September 2026). Google reports close variants with one combined figure, so a family total counts each distinct monthly series once, and every query's own figure is listed. A second measured source, a DataForSEO pull of the same Google Ads data for the US and 27 states, is kept as a cross-check: a record per place counts the queries on which the two agree. Free Google sources add, for the head queries with enough data, Google Trends interest by state over the last 12 months (a relative index, never mixed with counts) and the Google Autocomplete suggestions for the head queries whose suggestions pass the filter (only suggestions in which every word is a common word, so no firm, brand, person or place is named). Where our firm census holds 100 or more firm websites for a practice in a state, a ratio gives the searches a month per firm website, with both inputs.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/search-demand/>
