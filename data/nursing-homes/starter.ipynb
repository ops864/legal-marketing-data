For every US county with a Medicare- or Medicaid-certified nursing home, every state and territory, and the US: how many certified nursing homes and beds there are, how many times the Centers for Medicare & Medicaid Services (CMS) fined them in the last three years and for how many dollars, how often and for how many days it denied payment, and how many health citations inspectors wrote at the actual-harm level or worse (scope and severity G to L), at the immediate-jeopardy level (J to L) and under the abuse, neglect and exploitation tags (F600 to F610), plus the share of rated homes with one star overall.
