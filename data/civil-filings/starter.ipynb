# Federal civil filings by nature of suit and district, 12-month periods since 2018

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23720009](https://doi.org/10.5281/zenodo.23720009). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many civil cases were filed in each of the 94 U.S. district courts and nationwide, by nature of suit, from the Administrative Office of the U.S. Courts' Statistical Tables for the Federal Judiciary: Table C-3 (the 94 districts by case type: contract, real property, torts, civil rights, prisoner petitions, labor, Social Security, intellectual property and more, U.S. cases and private cases) and Table C-2 (the nation by about 100 natures of suit, such as product liability, medical malpractice, employment, the Fair Labor Standards Act, ERISA and consumer credit, with the basis of jurisdiction). One record per district and case type for the latest 12 months, with the year before and the history of every earlier table, and one record per nature of suit nationwide. In every table the 94 districts add up exactly to the national Total row, every circuit adds up to its districts and every nature-of-suit group adds up to the rows under it. Counts only: no lawyer or firm is listed or ranked.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/federal-civil-filings-by-type/>
