# The new-lawyer pipeline by jurisdiction: bar exam pass rates and new admissions, 2025

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23962982](https://doi.org/10.5281/zenodo.23962982). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

For every US jurisdiction that reports to the National Conference of Bar Examiners (the 50 states, DC, Guam, the Northern Mariana Islands, Palau, Puerto Rico and the Virgin Islands): the overall 2025 bar exam pass rate, how many new lawyers were admitted in 2025 (by examination, by transferred UBE score and on motion), how admissions by exam or transferred score changed since 2021, and how much of the total came from transferred scores and from motion. Every figure is derived from NCBE's published tables (a rate, a sum, a change, a share) and credited to NCBE with a link to the table it comes from. Where a jurisdiction shows a single admission pathway, the admitted figure is the count NCBE prints for that pathway (The Bar Examiner, statistics for 2025). Each jurisdiction's rows add up to NCBE's own totals row before anything is published.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/bar-pipeline/>
