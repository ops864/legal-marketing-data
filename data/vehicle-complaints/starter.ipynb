# Vehicle safety complaints by state, make and model year (NHTSA complaint file)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23964283](https://doi.org/10.5281/zenodo.23964283). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Safety complaints about cars, trucks, SUVs, motorcycles and RVs that owners filed with the National Highway Traffic Safety Administration (NHTSA), counted from its Office of Defects Investigation complaint file for the ten most recent model years, by the complainant's state, the make, the model year and the part of the vehicle named (engine, brakes, air bags, steering and the other NHTSA component groups). Each record gives the complaints, how many report a crash, a fire, an injury or a death, the people reported injured and killed, the complaints filed in the last twelve months, and how many concern a make, model and year that has a safety recall (and a recall of the same system), joined from NHTSA's recall file. State records list every make; make records add the model year by component table and each model's counts; national make records add each model by model year and the number of recall campaigns. A complaint is counted once however many parts it names. Only counts are kept: no narrative, VIN, city, name or complaint number.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/vehicle-complaints/>
