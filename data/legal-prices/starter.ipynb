# Legal-services prices and jobs, monthly (BLS CPI, PPI and CES)

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116042](https://doi.org/10.5281/zenodo.24116042). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Month by month since 2017, what consumers pay for legal services (Consumer Price Index), what law offices receive for their work (Producer Price Index for offices of lawyers, overall and by type of work), and how many people work in legal services (Current Employment Statistics), each next to the all-items or services index where that helps. Every value is read from the Bureau of Labor Statistics' own database with its series id, base period, and a preliminary flag while BLS may still revise it. The consumer index measures a basket item, not a law firm's billing rate. Counts and index values only.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/legal-prices/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/legal-prices.csv"
URL = "https://probativegrowth.com/data/v1/legal-prices/legal-prices.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
