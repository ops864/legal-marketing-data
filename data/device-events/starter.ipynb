# Medical device adverse event reports by product code and quarter (FDA MAUDE via openFDA)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963431](https://doi.org/10.5281/zenodo.23963431). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Adverse event reports the FDA received about medical devices, counted by device product code and calendar quarter from 2025 Q1: reports, deaths, injuries, malfunctions and other, for the 300 product codes with the most reports and for all devices together. Counted with the openFDA device event API (licence CC0); no report row, narrative, patient or manufacturer is ever read, only counts. Reports are filed up to weeks after an event and are driven by litigation and recalls as well as by harm, so a count is not an incidence rate.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/device-events/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/device-events.csv"
URL = "https://probativegrowth.com/data/v1/device-events/device-events.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
