# Google search-ad costs for lawyer searches by practice area and state: top-of-page bid bands and competition (Google Ads Keyword Planner)

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23851278](https://doi.org/10.5281/zenodo.23851278). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

What advertisers bid to show at the top of Google for lawyer searches, practice by practice. For 86 query families (English and Spanish) in the whole US, all 50 states and DC, the file gives the top-of-page bid band of each family's English and Spanish head query (Google Keyword Planner's low and high top-of-page bid estimates: the range advertisers have historically bid to show above the results), the competition level and index, plus the family's median low and high bids across its queries. Figures come from Google Ads Keyword Planner (October 2025 to September 2026) and are shown as bands with that period; for the US and 27 states, a DataForSEO pull of the same Google Ads data agrees on every bid band it shares. For the US, families are grouped into demand tiers by searches a month with the median high bid of each tier. Our own anonymized benchmark of what law-firm clicks actually cost is linked as a separate measure and never mixed into these figures.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/ad-costs/>
