Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23964225](https://doi.org/10.5281/zenodo.23964225). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

The unemployment rate, number unemployed, employed and labor force for every county and state, from the Bureau of Labor Statistics' Local Area Unemployment Statistics (LAUS) program: the annual average for each year 2015 to 2025 and the latest 13 months. Each county and state record carries the latest annual average, the change from the year before and since 2015, and the latest month (preliminary until BLS revises it). County figures are model-based estimates that BLS controls to its state totals; the build checks that county employment, unemployment and labor force add up to the state's. Alaska appears with its current census-area codes, and Connecticut by its planning regions.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/unemployment/>
