# Truck and bus crashes by county, state and year (FMCSA crash file)

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116060](https://doi.org/10.5281/zenodo.24116060). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Crashes involving a commercial truck or bus that states reported to the Federal Motor Carrier Safety Administration (FMCSA), counted from its Motor Carrier Management Information System (MCMIS) crash file for every county, state and territory and the US over the last ten complete calendar years. Each record gives, for the latest year, the crashes, the fatal crashes, the crashes with injuries, the crashes where a vehicle was towed away, the crashes involving a bus, the deaths and the injuries, with the whole ten-year history and the ten-year totals. A crash is one state crash report, counted once however many trucks and buses it involved, and its deaths and injuries are counted once. State records add the number of active carriers based in the state, from the FMCSA Company Census File. Counties with no reported crash are listed with zeros, and reports without a usable county code are kept in a 'county not given' row for their state, so counties always add up to the state.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/truck-crashes/>
