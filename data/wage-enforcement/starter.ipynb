# Wage-and-hour and OSHA enforcement by county, state, act and industry (U.S. Department of Labor)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23898146](https://doi.org/10.5281/zenodo.23898146). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

What the U.S. Department of Labor found when it enforced wage and safety laws, counted by county, state and territory and for the US over the last ten complete federal fiscal years. From the Wage and Hour Division (WHD): concluded cases, cases with violations, back wages found owed, employees owed back wages, civil money penalties, Fair Labor Standards Act overtime and minimum wage back wages, and child labor violations. From the Occupational Safety and Health Administration (OSHA), federal and state-plan offices: inspections (and how many followed a complaint, a referral, an accident or a fatality or catastrophe), violations cited (serious, willful, repeat), and penalties. State and US records add one record per law WHD enforces (FLSA, FMLA, Davis-Bacon, the Service Contract Act, H-2A, H-2B, H-1B, MSPA and others) and one per industry sector. Only counts and sums are published: no employer name, address or case number, and back wages only where 3 or more employees were owed.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/wage-enforcement/>
