Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963186](https://doi.org/10.5281/zenodo.23963186). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Reports that consumers made to the U.S. Federal Trade Commission's Consumer Sentinel Network in 2024, by state, DC and Puerto Rico, from the FTC's Consumer Sentinel Network Data Book 2024. Four records per jurisdiction: fraud and other reports and identity theft reports (each with the count, the rate per 100,000 people and the FTC's rank among the 50 states), imposter-scam reports (the share of the state's reports that the FTC prints, with a count and a rate derived from it, flagged as rounded) and fraud reports with the share reporting a loss, the total dollars reported lost and the median loss. Four national records give the whole country. A report is what a consumer or partner told the FTC, not a finding; the state tables leave out reports from state-specific data contributors and reports with no state.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/consumer-sentinel/>
