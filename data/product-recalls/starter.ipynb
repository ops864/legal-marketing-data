# Consumer product recalls by year and hazard (CPSC recall notices)

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116056](https://doi.org/10.5281/zenodo.24116056). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Recalls announced by the U.S. Consumer Product Safety Commission (CPSC) since 2010, counted for the whole period, for each recall year and for each of ten hazard classes (fire and burn, falls and tip-overs, laceration, choking and suffocation, ingestion, entrapment and pinch injuries, electric shock, poisoning and chemicals, crashes and collisions, drowning, plus a class for the rest). Each count comes with the number of recalls whose notice reports an injury and the number that report a death, parsed from the notice's own incident text, and with the units the notices give. A recall that names several hazards counts once in each class and once in the totals. The counts come from CPSC's recall API; no firm names, contacts or narratives are kept.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/product-recalls/>
