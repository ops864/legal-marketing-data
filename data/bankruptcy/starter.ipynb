# Bankruptcy filings by chapter, court district, state and county, quarterly (U.S. Courts)

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23739952](https://doi.org/10.5281/zenodo.23739952). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Bankruptcy cases commenced in the 12 months ending on each quarter-end, from the U.S. Courts' Tables F-5A (by county and chapter), F-2 (by court district, chapter and business or nonbusiness debt) and F (cases filed, terminated and pending). One record per area: the United States, every state, DC and territory, 93 court districts and about 3,100 counties. Each gives the latest period's cases by Chapter 7, 11, 13 and other chapters, the business and nonbusiness split, the change on the year before, filings per 1,000 residents (Census 2025 population estimates) and the court's figures for the six earlier periods. Every table is checked against the court's own totals row. Counts of cases, not outcomes: a filing is not a discharge.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/bankruptcy-filings-by-county/>
