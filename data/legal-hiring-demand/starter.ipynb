What the roles around a law firm's growth work pay and how many people hold them, and how hiring is moving. Annual pay bands and headcounts (10th, 25th, median, 75th and 90th percentile and mean annual wage) for marketing managers, marketing and market-research specialists, public relations specialists, writers, web designers, web developers, receptionists and customer service representatives from the Bureau of Labor Statistics' Occupational Employment and Wage Statistics (OEWS): nationally within legal services, nationally across all industries, and in every state and DC. Monthly series add legal services employment and average hourly earnings, job openings, hires and quits in professional and business services, and each state's average job-openings and hires rates over the latest year for all nonfarm employers.
