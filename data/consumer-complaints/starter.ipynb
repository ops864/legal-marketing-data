# CFPB complaints about debt collection, credit reporting and debt management, by state and month

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963181](https://doi.org/10.5281/zenodo.23963181). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Complaints that consumers filed with the Consumer Financial Protection Bureau (CFPB) about debt collection, credit reporting and credit repair, and debt or credit management, counted by state, month, product and issue from January 2018, with the last 12 months set against the 12 before and per 100,000 residents. One record per state, DC, Puerto Rico and territory for each of the three product groups, with the share of complaints the company answered on time and the issue people raised most. The counts come from the CFPB's own Consumer Complaint Database search API, as aggregates, so no complaint, narrative, company name or ZIP code is ever copied; for every month the issue counts add up to the bureau's own count for each product group. The bureau changed its credit-reporting product name twice; every name is combined into one group. Complaints are consumers' own reports, not findings of any violation.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/consumer-finance-complaints-by-state/>
