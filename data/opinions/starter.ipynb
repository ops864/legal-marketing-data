# Legal-marketing ethics opinions, 2004–2026

Starter notebook, version 1.1.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23734859](https://doi.org/10.5281/zenodo.23734859). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.1.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

An index of US state and ABA ethics opinions, committee guidelines and court rulings on lawyer advertising and marketing, from 2004 to 2026 (most from 2015 on, plus earlier landmarks): lead generators, pay-per-lead, reviews, testimonials, trade names, websites, social media, chatbots and AI tools, competitor keywords and more. Version 1.1 adds an AI sub-index: 14 AI opinions coded on six marketing questions (chatbot disclosure, AI-made ads, intake and solicitation, superiority claims, confidentiality in marketing tools, and fees). Each record gives the issuing body, number, date, topics, a one-sentence holding in our words, a practical step, whether it is still in force, the official link, an exact short quote and the date read. It is not the opinions' full text.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/opinions/>
