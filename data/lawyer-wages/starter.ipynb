```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/lawyer-wages.csv"
URL = "https://probativegrowth.com/data/v1/lawyer-wages/lawyer-wages.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
