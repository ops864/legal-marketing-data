# Texas employers without workers' compensation coverage (non-subscribers), by county

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116061](https://doi.org/10.5281/zenodo.24116061). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Texas does not require most private employers to carry workers' compensation insurance. Employers that opt out file a notice of no coverage (DWC Form-005) with the state each year. This dataset counts those filings from October 2024 to September 2026: for the state, and for each county with at least five non-subscribing employers, the employers with a location in the county, their locations and the filings by year. Counties with fewer than five are combined. Employers are businesses: no company names or addresses are kept. A ZIP code is placed in the county that holds most of its land.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/tx-nonsubscribers/>
