# Federal sentencing by district and offense: individuals sentenced, prison rates and sentence lengths, fiscal years 2021 to 2025

Starter notebook, version 1.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24022955](https://doi.org/10.5281/zenodo.24022955). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

The U.S. Sentencing Commission's counts of individuals sentenced in federal court for the nation and each federal judicial district, overall and by type of offense (drug trafficking, firearms, fraud, immigration and 26 others): how many were sentenced, the share sentenced to prison, the average and median sentence in months (prison plus alternative confinement, probation as zero), and the shares convicted by guilty plea and after trial. Built from the Commission's public datafiles without individual identifiers, reduced to cell counts; each district and offense listed had 10 or more individuals in the latest year.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/federal-sentencing/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/federal-sentencing.csv"
URL = "https://probativegrowth.com/data/v1/federal-sentencing/federal-sentencing.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
