Headline counts about the US legal market, each computed from one or more of Probative's open datasets and published as a quarterly edition: counties with no law office, Spanish speakers against Spanish-speaking lawyers, ad-rule waiting periods and filing duties, truck-crash deaths, workplace charges, the immigration court backlog, veterans' representation, deadlines to sue, legal-services prices, who earns the fees, mass-tort dockets, disability hearing waits, bankruptcy filings, nursing-home fines and unpaid wages. ## Load the data / Cargar los datos
