# Eviction filings by county and year (LSC Civil Court Data Initiative)

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963459](https://doi.org/10.5281/zenodo.23963459). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Eviction filings (cases) by county and month from the Legal Services Corporation's Civil Court Data Initiative Eviction Tracker, which collects residential eviction filings from state and local courts: 1,464 counties and 34 states and territories, January 2016 through September 2025. Each record carries the filings of every full calendar year the county reports, the last twelve months where all twelve are present, filings per 100 renter households, the poverty rate and the latest month against the county's own historical average. Counties carry the Census FIPS code. LSC's data are released under CC0 1.0. The states and territories in the file are the ones LSC reports; the US record totals a year or month only where every one of them reports it.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/evictions/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.
