# State trial-court case filings by county, court level and case type: Texas, Washington, Maryland, Indiana, Wisconsin, Virginia, Pennsylvania, Oregon, West Virginia, Nebraska, Maine, California, Louisiana and Hawaii

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23964136](https://doi.org/10.5281/zenodo.23964136). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many cases are filed in state trial courts, by county, court level and case group, from each state court system's own published statistics. Texas (Office of Court Administration): district, constitutional county and statutory county courts, fiscal years 2023–24 and 2024–25, with district-court filings by case type. Washington (Administrative Office of the Courts): superior courts and district and municipal courts (criminal, civil, domestic, probate, juvenile, traffic, DUI, protection orders, small claims), calendar year 2025. Maryland (Judiciary data dashboard): circuit-court filings by case type and District Court filings (landlord-tenant, civil, criminal, DUI, traffic citations, peace orders), fiscal years 2012 to 2025. Indiana (Office of Judicial Administration): circuit and superior court new filings by Rule 8 case type (felony, misdemeanor, tort, collection, small claims, eviction, foreclosure, domestic relations, estates, juvenile), calendar year 2025 (provisional) beside 2024. Wisconsin (Director of State Courts Office): circuit-court cases opened by category, felony to juvenile, 2020 to 2025. Virginia (Office of the Executive Secretary): circuit, general district and juvenile and domestic relations district courts by locality (county or independent city), calendar year 2025. Pennsylvania (Administrative Office of Pennsylvania Courts): courts of common pleas in all 67 counties, civil cases by type (motor-vehicle and other torts, mass torts, malpractice, contract and debt, landlord-tenant, foreclosure) and criminal, protection from abuse, support, custody, divorce, juvenile, adoption and guardianship cases, calendar year 2024. Oregon (Oregon Judicial Department): circuit courts in all 36 counties, calendar year 2025. West Virginia (Supreme Court of Appeals): circuit, family and magistrate courts in all 55 counties, calendar year 2025. Nebraska (Judicial Branch): county and district courts in all 93 counties and the separate juvenile courts of three of them, fiscal year 2024–25. Maine (Judicial Branch): superior courts, unified criminal dockets and district courts in all 16 counties, fiscal years 2021–22 to 2025–26. California (Judicial Council): superior courts in all 58 counties (motor-vehicle and other injury, limited civil, eviction, small claims, felony, misdemeanor, traffic, marital, parentage, domestic violence, child support, juvenile delinquency and dependency, guardianship, other probate, mental health), fiscal year 2024–25. Louisiana (Supreme Court of Louisiana, Judicial Administrator's Office): district courts in 61 of the state's 64 parishes (civil, criminal, traffic and juvenile cases), calendar year 2025 beside 2024 where the 2024 report's row matches. Hawaii (Hawaii State Judiciary): circuit courts in all four circuits, each mapped to its county (motor-vehicle and other injury, contract, foreclosure, probate proceedings, circuit criminal cases), fiscal year 2024–25. Every record keeps the state's own category names and the numbers added into each group (components), and carries a harmonized group and the practice areas it bears on, so a county's filings can be read as demand for personal injury, family, landlord-tenant, small claims and debt, probate, criminal and DUI, juvenile and traffic work. In every source the counties add up to the state's own printed total where it prints one (Texas, Washington, Indiana, Wisconsin, Virginia, Pennsylvania, Oregon, West Virginia, Nebraska, Hawaii), Louisiana's parish records are the parishes whose rows add up to the printed district totals and its state row is the state's printed total, and the state rows for Maryland are the sum of the counties. Every record carries its FIPS code (the state's on a state row, the county's on a county row) so it joins Census and other county data. Counts only: no lawyer or firm is listed or ranked.
