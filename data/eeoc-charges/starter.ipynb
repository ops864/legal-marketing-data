# Employment discrimination charges to the EEOC, by state and basis, FY 2009 to FY 2025

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23898101](https://doi.org/10.5281/zenodo.23898101). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Charges of workplace discrimination received by the U.S. Equal Employment Opportunity Commission (EEOC), by state, DC, Puerto Rico and territory, fiscal year and basis (race, sex, national origin, religion, color, retaliation, age, disability, equal pay, genetic information and pregnancy under the Pregnant Workers Fairness Act), from the EEOC's Tables E1b and E1a. One record per jurisdiction and basis for the latest fiscal year, each with its count, its share of the state's charges, the rate per 100,000 private-sector workers (BLS Quarterly Census of Employment and Wages, from fiscal year 2014) and the whole FY 2009 to FY 2025 history. A charge can name more than one basis, so bases add up to more than the total, and charges filed only with a state or local fair-employment agency are not counted.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/eeoc-charges/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/eeoc-charges.csv"
URL = "https://probativegrowth.com/data/v1/eeoc-charges/eeoc-charges.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
