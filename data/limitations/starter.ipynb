# Statutes of limitations by state: personal injury, medical malpractice, wrongful death, product liability and property damage

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963806](https://doi.org/10.5281/zenodo.23963806). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

The filing deadline the statute sets for five kinds of civil claims in every US state and DC: personal injury, medical malpractice, wrongful death, product liability and property damage. Each record gives the period in the statute's own words, the citation, a link to the state's official code site, a verbatim quote, and the date read; where the same text says so, it adds whether the period runs from discovery and any outer limit (statute of repose). Some cells come from a court opinion that quotes or applies the statute; they are marked S (secondary) and name the court, case and date. No calculator and no deadline arithmetic: tolling and exceptions apply.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/statute-of-limitations-by-state/>
