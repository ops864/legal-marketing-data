# Legal need by county: who may need a lawyer, from the American Community Survey

Starter notebook, version 1.1.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23845845](https://doi.org/10.5281/zenodo.23845845). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.1.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

The people in every U.S. county who may need a lawyer, counted from the Census Bureau's American Community Survey 2020 to 2024 5-year estimates: residents aged 65 and over (and 75 and over and 85 and over), people aged 65 or over living alone and those below the poverty level, residents who are not U.S. citizens, foreign-born residents who arrived since 2010, residents who speak English less than very well (and Spanish speakers among them), divorced residents (and, for states, people divorced in the last year), veterans and veterans with a VA disability rating, residents with a disability, households on SSI, renter households and renters paying 30 or 50 percent of income on rent, residents without health insurance, workers who drive to work and median household income. One record per county (3,144, in the 50 states and DC), per state and for the US, each measure with its 90% margin of error and its share of the population it is counted in.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/legal-need/>
