Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23964168](https://doi.org/10.5281/zenodo.23964168). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Toxic chemicals that industrial and federal facilities reported releasing on site to the Environmental Protection Agency's Toxics Release Inventory (TRI) in reporting year 2024, for 2,437 counties with a reporting facility, 56 states and territories and the US: 21,482 facilities and 2,842,308,160 pounds in all (530,322,130 to air, 188,664,266 to water, 2,123,321,764 to land), with the pounds of known or suspected carcinogens, PFAS, ethylene oxide, benzene and lead, and the on-site total of each of the earlier years kept (2020, 2021, 2022, 2023). State records list the three largest parent companies. Counties carry the Census FIPS code. Dioxin-like compounds, which facilities report in grams, are converted to pounds. TRI counts reported releases by listed chemicals from covered facilities; it is not a measure of exposure or of harm.
