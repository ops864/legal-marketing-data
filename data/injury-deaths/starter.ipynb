Deaths from drug overdose, suicide, homicide and firearm injury in CDC's Mapping Injury, Overdose, and Violence files, for the nation, each state and DC and each county, by calendar year and for the trailing twelve months: deaths and the rate per 100,000 residents. National and state records add unintentional-injury deaths from the NCHS leading-causes file, 1999 to 2017. Counts of 1 to 9 stay in the range CDC prints.

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/injury-deaths.csv"
URL = "https://probativegrowth.com/data/v1/injury-deaths/injury-deaths.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
