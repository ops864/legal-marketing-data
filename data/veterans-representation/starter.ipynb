# VA-accredited attorneys and claims agents per veteran, by state, with county counts and Board of Veterans' Appeals outcomes by type of representation

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23740008](https://doi.org/10.5281/zenodo.23740008). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many attorneys and claims agents the U.S. Department of Veterans Affairs has accredited to help with benefit claims, counted by the state (and, in the county file, the county) of each person's business address, against the VA's own projection of living veterans (VetPop2023, rounded to the nearest 1,000). One record per state, DC, Puerto Rico and territory, one for the United States and one for people whose list entry shows no US state, each with attorneys, claims agents, veterans per accredited attorney, accredited attorneys and agents per 100,000 veterans, the number accredited in the last 12 months and how many counties have none. Records for the Board of Veterans' Appeals add its decisions in fiscal year 2024 by outcome and by type of representative. The VA's lists name every person; only counts are read, and no name, phone number, registration number or ZIP code is published.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/veterans-representation/>
