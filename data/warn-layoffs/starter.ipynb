# WARN Act layoff and closure notices by county: California and Texas

Starter notebook, version 2.0.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24116063](https://doi.org/10.5281/zenodo.24116063). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

How many mass-layoff and plant-closure notices employers filed under the Worker Adjustment and Retraining Notification (WARN) Act, and how many employees they covered, for every county in the states whose workforce agencies publish the notices as files: California (July 2023 on, from the Employment Development Department's reports) and Texas (2019 on, from the Texas Workforce Commission's open-data table). Each county, state and county-not-given record carries the last 12 months, the last complete calendar year, the total since the file starts, and the month-by-month series. Only counts are kept: no employer name, address or contact. Notices are dated by the day the state agency received them.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/warn-layoffs/>
