# Serious drug adverse event reports by active ingredient (openFDA FAERS)

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23752994](https://doi.org/10.5281/zenodo.23752994). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Adverse event reports about drugs that reached the FDA Adverse Event Reporting System (FAERS) and are served by openFDA, received 2024Q3 to 2026Q2: 2,649,395 reports in all, 1,430,033 serious and 197,378 with a death outcome. One record per active ingredient for the 312 ingredients with the most reports plus every ingredient named in an open multidistrict litigation (31 of them tagged with their MDL numbers): reports that list the ingredient, how many were serious, how many reported a death outcome, the same counts for each of the last eight quarters, and the five most reported reactions. National, one shard. A report names a drug and an event; it does not show that the drug caused the event, and FAERS holds duplicates, follow-ups and reports from outside the United States.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/drug-events/>
