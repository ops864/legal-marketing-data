# Unwanted-call and text-message complaints to the FCC, by state and month

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23964125](https://doi.org/10.5281/zenodo.23964125). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Complaints about unwanted calls, robocalls and telemarketing that people filed with the Federal Communications Commission (FCC), counted by state, month, issue and method (text message, prerecorded voice, live voice and others) from January 2015, with the last 12 months set against the 12 before and per 100,000 residents. One record per state, DC, Puerto Rico and territory for text-message complaints and one for phone-call complaints. The counts come from the FCC's open Consumer Complaints Data, grouped on the FCC's server so no complaint, caller number or free text is ever copied, and the state rows add up to a separate FCC-side count for every month. Complaints are self-reported and unverified, not findings of any violation.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/robocall-complaints/>
