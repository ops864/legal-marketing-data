# How long federal civil cases take: outcomes, class actions and pro se share by district, nature of suit and year

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23898111](https://doi.org/10.5281/zenodo.23898111). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Federal civil cases from the Federal Judicial Center's Integrated Database (IDB), counted by district, nature-of-suit group and fiscal year: cases filed, the share alleging a class action, the share with a pro se plaintiff, cases terminated, the median days from filing to disposition, the trial share, the median days to trial, and how terminated cases ended (settled, voluntarily dismissed, dismissed, default or consent, motion, trial, other, moved out). One record per district and group for the latest three fiscal years, each with its yearly series, plus a national record per nature-of-suit code. Only aggregates are published: no case-level row, party, docket number or judge. Filings re-add to the U.S. Courts' own totals within half a percent.
