# US law firm websites: practices, Spanish, intake, by state, county and practice area

Starter notebook, version 3.2.1. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24148921](https://doi.org/10.5281/zenodo.24148921). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 3.2.1. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

US law firms, state by state and county by county, from two kinds of open source: law firms' own public websites, and official state registries published as open data for reuse (New York's attorney registrations and business entities, and the business registries of Colorado, Oregon, Connecticut, Texas and Pennsylvania). For each state, county and practice area we count the firms found, how many have a website we read, and, of those websites, the shares that show Spanish-language pages, a 24/7 intake statement, a free consultation or no fee unless they win, a chat or online booking tool, legal structured data (schema.org LegalService or Attorney), review markup or an llms.txt file, with the median number of pages a firm has about the practice. Every row compares the firms found with the employer law offices the Census Bureau's County Business Patterns counts. A state is called a census or a registry list only when that reaches 0.70; until then the rows are a sample. These counts are CC BY 4.0. The firm-level rows behind them (name as the firm or registry gives it, offices by county and the yes-or-no facts, each linked to the page it was read on with the date) are published separately under the Open Database License (ODbL) 1.0, because the list of sites comes partly from OpenStreetMap (© OpenStreetMap contributors), and are shown on the /firms/ pages. No firm is rated, scored or ranked. The firm rows hold no street address, phone number or email address: a page address that contains a phone-shaped number or a street-address shape is left out, and a name that contains one is counted and not listed. Firm names can include a lawyer's name when that is the firm's registered or website name; a row that is only one person's name is counted and not listed. Website sources: Common Crawl (its web graph and the Web Data Commons structured-data extract), OpenStreetMap and Wikidata.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/firm-web-census/>
