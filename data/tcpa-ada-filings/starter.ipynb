# TCPA and ADA federal lawsuits filed, by district and fiscal year

Starter notebook, version 1.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.24187571](https://doi.org/10.5281/zenodo.24187571). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 1.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Civil cases filed in each of the 94 federal district courts in each federal fiscal year under nature-of-suit code 485 (Telephone Consumer Protection Act) and code 446 (Americans with Disabilities Act, other: cases outside employment under the ADA sections on public entities and public accommodations, not only website cases), with the cases inside a multidistrict litigation counted apart and every civil case the district received that year. Counted from the Federal Judicial Center's civil case file, with each code's definition quoted from the U.S. Courts. Counts only: no party, lawyer or case number.

Dataset page / Página del conjunto: <https://probativegrowth.com/data/tcpa-ada-filings/>

## Load the data / Cargar los datos

On Kaggle this reads the attached copy; anywhere else it reads the CSV from the source site. / En Kaggle lee la copia adjunta; en cualquier otro lugar lee el CSV del sitio de origen.

```python
import os
import pandas as pd

LOCAL = "/kaggle/input/legal-marketing-open-data-us/tcpa-ada-filings.csv"
URL = "https://probativegrowth.com/data/v1/tcpa-ada-filings/tcpa-ada-filings.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL, low_memory=False)
df.shape
```
