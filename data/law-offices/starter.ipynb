# Law offices, jobs and pay by county, with solo-practice counts: BLS QCEW, Census CBP and Nonemployer Statistics

Starter notebook, version 2.0.0. Published by [Probative Growth](https://probativegrowth.com/), a marketing agency for US law firms. DOI [10.5281/zenodo.23963687](https://doi.org/10.5281/zenodo.23963687). CC BY 4.0. Not legal advice.

Cuaderno inicial, versión 2.0.0. Publicado por Probative Growth, una agencia de marketing para firmas de abogados de EE. UU. No es asesoría legal.

Private offices of lawyers (NAICS 541110) in every U.S. county, every state and the nation. For each area: establishments, March employment and average weekly pay from the Bureau of Labor Statistics' Quarterly Census of Employment and Wages (the latest quarter, with the seven before it); employer offices from the Census Bureau's County Business Patterns 2023 exactly as first published; firms with no employees in legal services from Nonemployer Statistics 2024; and the 2025 Census population estimate, so each area has offices and solo firms per 10,000 residents. Employment and pay are given exactly as BLS publishes them for each area, never estimated. An office is an establishment, not a lawyer.

Dataset page / Página del conjunto: <https://probativegrowth.com/research/law-firm-competition-by-county/>
